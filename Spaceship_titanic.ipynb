```python
train_data= train_data.drop(columns="PassengerId")
train_data= train_data.drop(columns="Name")
train_data.head()
```

```python
test_data= test_data.drop(columns="PassengerId")
test_data= test_data.drop(columns="Name")
test_data.head()
```

## Handling Missing Values in train and test data

```python
train_data.isnull().mean()*100
```

```python
#For home planet
train_data['HomePlanet'] = train_data['HomePlanet'].fillna("U")
test_data['HomePlanet'] = test_data['HomePlanet'].fillna("U")
```

```python
train_data["HomePlanet"].isnull().mean()*100
```

```python
# For cabin
train_data['Cabin'] = train_data['Cabin'].fillna('U')
train_data['Deck'] = train_data['Cabin'].apply(lambda x: x[0])
train_data
```

```python
# Handling missing values in cyrosleep
amenities = ['RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']
train_data['TotalShopping'] = train_data[amenities].sum(axis=1)
train_data.loc[train_data['CryoSleep'].isnull() & (train_data['TotalShopping'] > 0), 'CryoSleep'] = False
train_data.loc[train_data['CryoSleep'].isnull() & (train_data['TotalShopping'] == 0), 'CryoSleep'] = True
```

```python
# Handling missing values in amenities
for col in amenities:
    train_data.loc[train_data['CryoSleep'] == True, col] = train_data[col].fillna(0)
```

```python
# Handling missing values in destination
destination_mode = train_data['Destination'].mode()[0]
train_data['Destination'] = train_data['Destination'].fillna(destination_mode)
```

```python
# Handling missing values in age by imputing median in NaN
from sklearn.impute import SimpleImputer
sim_imp= SimpleImputer(strategy="median")
train_data['Age']= pd.DataFrame(sim_imp.fit_transform(train_data[['Age']]))
```

```python
#For missing values in VIP we can check whether they have used luxury amentites or not
train_data.loc[train_data['VIP'].isnull() & (train_data['TotalShopping'] > 0), 'VIP'] = True
train_data.loc[train_data['VIP'].isnull() & (train_data['TotalShopping'] == 0), 'VIP'] = False
```

```python
# To Handle missing values in amenitites 
for ser in amenities:
   train_data[amenities] = train_data[amenities].fillna(0)
```

```python
train_data.isnull().mean()*100
```

## Converting all string data to int/float

```python
train_data['HomePlanet'].unique()
```

```python
# Using Encoder for Homeplanet
train_data = pd.get_dummies(train_data, columns=["HomePlanet"], prefix="HomePlanet", dtype=int)
```

```python
train_data
```

```python
## Using ordinal encoder for these cols
cols=['CryoSleep','VIP','Transported']
from sklearn.preprocessing import OrdinalEncoder
for col in cols:
    cat= list(train_data[col].unique())
    ord_enc= OrdinalEncoder(categories=[cat])
    train_data[col]= pd.DataFrame(ord_enc.fit_transform(train_data[[col]]))
```

```python
train_data
```

```python
train_data['Destination'].unique()
```

```python
train_data = pd.get_dummies(train_data, columns=['Destination'], prefix='Destination', dtype=int)
train_data
```

```python
## For encoding cabin, split it into three parts
train_data[['Deck', 'Num', 'Side']] = train_data['Cabin'].str.split('/', expand=True)
train_data['Num'] = pd.to_numeric(train_data['Num'], errors='coerce')
train_data['Num'] = train_data['Num'].fillna(train_data['Num'].median())
```

```python
train_data = pd.get_dummies(train_data, columns=['Deck'], prefix='Deck', dtype=int)
train_data['Side'] = train_data['Side'].map({'P': 0, 'S': 1}).fillna(-1).astype(int)
train_data = train_data.drop(columns=['Cabin'])
```
