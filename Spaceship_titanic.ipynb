```python
test_data= pd.read_csv("/kaggle/input/competitions/spaceship-titanic/test.csv")
test_data
```

```python
train_data= train_data.drop(columns="PassengerId")
train_data= train_data.drop(columns="Name")
train_data.head()
```

```python
test_data= test_data.drop(columns="PassengerId")
test_data= test_data.drop(columns="Name")
test_data.head()
```

## Handling Missing Values in train and test data

```python
train_data.isnull().mean()*100
```

```python
#For home planet
train_data['HomePlanet'] = train_data['HomePlanet'].fillna("U")
test_data['HomePlanet'] = test_data['HomePlanet'].fillna("U")
```

```python
train_data["HomePlanet"].isnull().mean()*100
```

```python
# For cabin
train_data['Cabin'] = train_data['Cabin'].fillna('U')
train_data['Deck'] = train_data['Cabin'].apply(lambda x: x[0])
train_data
```

```python
# Handling missing values in cyrosleep
amenities = ['RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']
train_data['TotalShopping'] = train_data[amenities].sum(axis=1)
train_data.loc[train_data['CryoSleep'].isnull() & (train_data['TotalShopping'] > 0), 'CryoSleep'] = False
train_data.loc[train_data['CryoSleep'].isnull() & (train_data['TotalShopping'] == 0), 'CryoSleep'] = True
```

```python
# Handling missing values in amenities
for col in amenities:
    train_data.loc[train_data['CryoSleep'] == True, col] = train_data[col].fillna(0)
```

```python
# Handling missing values in destination
destination_mode = train_data['Destination'].mode()[0]
train_data['Destination'] = train_data['Destination'].fillna(destination_mode)
```

```python
# Handling missing values in age by imputing median in NaN
from sklearn.impute import SimpleImputer
sim_imp= SimpleImputer(strategy="median")
train_data['Age']= pd.DataFrame(sim_imp.fit_transform(train_data[['Age']]))
```

```python
#For missing values in VIP we can check whether they have used luxury amentites or not
train_data.loc[train_data['VIP'].isnull() & (train_data['TotalShopping'] > 0), 'VIP'] = True
train_data.loc[train_data['VIP'].isnull() & (train_data['TotalShopping'] == 0), 'VIP'] = False
```

```python
# To Handle missing values in amenitites 
for ser in amenities:
   train_data[amenities] = train_data[amenities].fillna(0)
```

```python
train_data.isnull().mean()*100
```

```python
train_data
```

## Converting all string data to int/float

```python
train_data['HomePlanet'].unique()
```

```python
# Using Encoder for Homeplanet
train_data = pd.get_dummies(train_data, columns=["HomePlanet"], prefix="HomePlanet", dtype=int)
```

```python
train_data
```

```python
## Using ordinal encoder for these cols
cols=['CryoSleep','VIP','Transported']
from sklearn.preprocessing import OrdinalEncoder
for col in cols:
    cat= list(train_data[col].unique())
    ord_enc= OrdinalEncoder(categories=[cat])
    train_data[col]= pd.DataFrame(ord_enc.fit_transform(train_data[[col]]))
```

```python
train_data
```

```python
train_data['Destination'].unique()
```

```python
train_data = pd.get_dummies(train_data, columns=['Destination'], prefix='Destination', dtype=int)
train_data
```

```python
## For encoding cabin, split it into three parts
train_data[['Deck', 'Num', 'Side']] = train_data['Cabin'].str.split('/', expand=True)
train_data['Num'] = pd.to_numeric(train_data['Num'], errors='coerce')
train_data['Num'] = train_data['Num'].fillna(train_data['Num'].median())
```

```python
train_data = pd.get_dummies(train_data, columns=['Deck'], prefix='Deck', dtype=int)
train_data['Side'] = train_data['Side'].map({'P': 0, 'S': 1}).fillna(-1).astype(int)
train_data = train_data.drop(columns=['Cabin'])
```

```python
train_data
```

## Doing feature scaling do scale down all features

```python
X= train_data.drop(columns = "Transported")
y= train_data["Transported"]
```

```python
from sklearn.preprocessing import StandardScaler
std_scaler = StandardScaler()
feature_columns = X.columns 
X_scaled = std_scaler.fit_transform(X)
X = pd.DataFrame(X_scaled, columns=feature_columns)
```

```python
X
```

## Doing Same things on test_data

```python
# 1. Deck Extraction & Missing Values Handling
test_data["Cabin"] = test_data["Cabin"].fillna("U")
test_data["Deck"] = test_data["Cabin"].apply(lambda x: x[0])

# 2. Total Amenities & CryoSleep Imputation
amenities = ["RoomService", "FoodCourt", "ShoppingMall", "Spa", "VRDeck"]
test_data["TotalShopping"] = test_data[amenities].sum(axis=1)

test_data.loc[test_data["CryoSleep"].isnull() & (test_data["TotalShopping"] > 0),"CryoSleep",] = False
test_data.loc[test_data["CryoSleep"].isnull() & (test_data["TotalShopping"] == 0),"CryoSleep",] = True

# Zero out amenities for passengers in CryoSleep
for col in amenities:
  test_data.loc[test_data["CryoSleep"] == True, col] = test_data[col].fillna(0)

# 3. Impute Destination with train_data mode
test_data["Destination"] = test_data["Destination"].fillna(destination_mode)

# 4. Impute Age using fitted SimpleImputer
test_data["Age"] = sim_imp.transform(test_data[["Age"]])

# 5. Impute VIP based on spending
test_data.loc[test_data["VIP"].isnull() & (test_data["TotalShopping"] > 0), "VIP"] = True
test_data.loc[test_data["VIP"].isnull() & (test_data["TotalShopping"] == 0), "VIP"] = False

# Fill remaining missing amenity values with 0
test_data[amenities] = test_data[amenities].fillna(0)

# Fill missing HomePlanet values with 'U' if not already done
test_data["HomePlanet"] = test_data["HomePlanet"].fillna("U")

# 6. One-Hot Encoding for HomePlanet
test_data = pd.get_dummies(
    test_data, columns=["HomePlanet"], prefix="HomePlanet", dtype=int
)

# Align columns in case a category was missing in test set
test_data = test_data.reindex(
    columns=[c for c in train_data.columns if c != "Transported"], fill_value=0
)

# 7. Encode Boolean Columns
cols_to_encode = ["CryoSleep", "VIP"]
for col in cols_to_encode:
  test_data[col] = test_data[col].astype(int)
    
# 8. Align columns to match X's exact structure before scaling
test_data = test_data.reindex(columns=feature_columns, fill_value=0)

# 9. Apply the fitted StandardScaler from Cell 30
X_test_scaled = std_scaler.transform(test_data)
test = pd.DataFrame(X_test_scaled, columns=feature_columns)
```

## Choosing and Training of model

```python
## Train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train,y_test= train_test_split(X,y, random_state=42, test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier
import xgboost as xgb
import lightgbm as lgb
from sklearn.model_selection import cross_val_score

models = {
    'lr': LogisticRegression(),
    'svm': SVC(),
    'dtree': DecisionTreeClassifier(),
    'rf': RandomForestClassifier(n_estimators=50),
    'xg': xgb.XGBClassifier(n_estimators=100),
    'gb': GradientBoostingClassifier(),
    'adb': AdaBoostClassifier(),
    'lgbm': lgb.LGBMClassifier()
}

for name, model in models.items():
    scores = cross_val_score(model, X_train, y_train, cv=5)
    print(f"{name} accuracy: {scores.mean() * 100:.2f}%")
```

```python
## Doing hyperparamter tuning for lightgbm
from lightgbm import LGBMClassifier
from sklearn.model_selection import RandomizedSearchCV
param_distributions = {
    'n_estimators': [50, 100, 200, 500],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'num_leaves': [15, 31, 50, 63],
    'max_depth': [-1, 5, 10, 15],
    'subsample': [0.6, 0.8, 1.0],
    'colsample_bytree': [0.6, 0.8, 1.0],
    'min_child_samples': [10, 20, 30]
}
lgb_model = LGBMClassifier(random_state=42)

random_search = RandomizedSearchCV(
    estimator=lgb_model,
    param_distributions=param_distributions,
    n_iter=20,       
    cv=5,              
    scoring='accuracy',
    random_state=42,
    n_jobs=-1,          
    verbose=1
)

random_search.fit(X_train, y_train)

print(f"Best Accuracy: {random_search.best_score_ * 100:.2f}%")
print("Best Parameters:")
print(random_search.best_params_)

best_lgb = random_search.best_estimator_
```
